```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\hp\Desktop\FOOD PRODUCTION PROJECT\data\processed\crop_production_clean.csv"
)

df.head()
```

```python
#Create a dataset for each crop
maize = df[df["crop"] == "Maize"].copy()
cassava = df[df["crop"] == "Cassava"].copy()
rice = df[df["crop"] == "Rice"].copy()
yams = df[df["crop"] == "Yams"].copy()
```

```python
#Sort by year
maize = maize.sort_values("year")
cassava = cassava.sort_values("year")
rice = rice.sort_values("year")
yams = yams.sort_values("year")
```

```python
#Reset the index
maize.reset_index(drop=True, inplace=True)
cassava.reset_index(drop=True, inplace=True)
rice.reset_index(drop=True, inplace=True)
yams.reset_index(drop=True, inplace=True)
```

```python
print("Maize:", maize.shape)
print("Cassava:", cassava.shape)
print("Rice:", rice.shape)
print("Yams:", yams.shape)
```

```python
for name, crop_df in {
    "Maize": maize,
    "Cassava": cassava,
    "Rice": rice,
    "Yams": yams
}.items():

    print(f"\n{name}")
    print("First year:", crop_df["year"].min())
    print("Last year :", crop_df["year"].max())
    print("Missing years:",
          crop_df["year"].diff().gt(1).sum())
```

```python
output_path = r"C:\Users\hp\Desktop\FOOD PRODUCTION PROJECT\data\processed"

maize.to_csv(f"{output_path}\\maize_forecast.csv", index=False)
cassava.to_csv(f"{output_path}\\cassava_forecast.csv", index=False)
rice.to_csv(f"{output_path}\\rice_forecast.csv", index=False)
yams.to_csv(f"{output_path}\\yams_forecast.csv", index=False)
```
