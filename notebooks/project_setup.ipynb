# Machine Learning Forecasting and Optimization of Crop Production and Food Prices in Ghana

## Project Overview

This project combines machine learning forecasting and operations research optimization to support agricultural decision-making in Ghana.

The aim is to forecast crop production and staple food prices, then use optimization techniques to recommend better resource allocation and planning decisions.

---

## Research Objectives

1. Forecast crop production for selected staple crops in Ghana.
2. Forecast staple food prices using historical and external factors.
3. Compare statistical and machine learning forecasting approaches.
4. Develop an optimization model for agricultural planning and/or food distribution.
5. Evaluate how predictive models can support decision-making.

---

## Research Questions

1. Can machine learning models accurately forecast crop production and food prices in Ghana?
2. Which forecasting approach performs best among statistical and machine learning models?
3. What factors contribute most to changes in agricultural production and food prices?
4. How can forecasts improve agricultural resource allocation decisions?
